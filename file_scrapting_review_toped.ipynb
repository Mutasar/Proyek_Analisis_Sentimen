```python
# ... (bagian import dan setup) ...

# 3. Tentukan URL Dasar Ulasan dan Parameter
# Ganti dengan URL halaman ulasan produk Tokopedia yang sebenarnya
# Anda harus menemukan URL yang mengarah ke halaman ulasan dan memiliki parameter halaman.
url_dasar_ulasan = "https://www.tokopedia.com/pentolban/review" # <--- GANTI INI!

# Tentukan jumlah halaman yang ingin discrape
num_pages_to_scrape = 5

# List untuk menyimpan data ulasan
data_ulasan = []

print(f"Memulai scraping dari URL dasar: {url_dasar_ulasan}")

for page in range(1, num_pages_to_scrape + 1):
    # Bangun URL untuk halaman saat ini
    # Sesuaikan format URL ini berdasarkan bagaimana Tokopedia menangani pagination
    # Contoh jika parameter halaman adalah '&page='
    url_saat_ini = f"{url_dasar_ulasan}&page={page}"
    print(f"Scraping halaman {page}: {url_saat_ini}")

    try:
        # Mengirim permintaan HTTP GET
        response = requests.get(url_saat_ini)
        response.raise_for_status() # Munculkan error untuk status kode yang buruk (4xx atau 5xx)

        # Parse konten HTML
        soup = BeautifulSoup(response.content, 'html.parser')

        # >>> PENTING: SESUAIKAN SELECTOR CSS INI <<<
        # Ganti dengan selector CSS yang tepat untuk kontainer ulasan di halaman ulasan Tokopedia
        review_containers = soup.select('SELECTOR_KONTAINER_ULASAN_DISINI') # <--- GANTI INI!

        if not review_containers:
            print(f"Tidak ditemukan elemen ulasan dengan selector yang diberikan di halaman {page}.")
            if page == 1:
                 print("Mungkin struktur HTML berubah, URL tidak tepat, atau tidak ada ulasan.")
            break # Berhenti jika tidak ada ulasan di halaman ini (mungkin sudah mencapai halaman terakhir)

        # Ekstrak data dari setiap ulasan yang ditemukan
        for review in review_containers:
            try:
                # >>> PENTING: SESUAIKAN SELECTOR CSS INI UNTUK DETAIL <<<
                # Ganti dengan selector CSS yang tepat untuk nama pengguna, rating, teks ulasan, dll.
                user_name_elem = review.select_one('SELECTOR_NAMA_PENGGUNA_DISINI') # <--- GANTI INI!
                rating_elem = review.select_one('SELECTOR_RATING_DISINI') # <--- GANTI INI!
                ulasan_elem = review.select_one('SELECTOR_TEKS_ULASAN_DISINI') # <--- GANTI INI!
                time_elem = review.select_one('SELECTOR_WAKTU_ULASAN_DISINI') # <--- GANTI INI!


                # Ambil teks dari elemen, atau 'N/A' jika tidak ditemukan
                user_name = user_name_elem.get_text(strip=True) if user_name_elem else 'N/A'
                # Untuk rating, Anda mungkin perlu mengurai atribut kelas atau data
                rating = rating_elem['aria-label'] if rating_elem and 'aria-label' in rating_elem.attrs else 'N/A' # Sesuaikan jika rating disimpan di atribut lain
                ulasan = ulasan_elem.get_text(strip=True) if ulasan_elem else 'N/A'
                time_str = time_elem.get_text(strip=True) if time_elem else 'N/A'


                data_ulasan.append({
                    'userName': user_name,
                    'rating': rating,
                    'ulasan': ulasan,
                    'waktu': time_str
                })
            except Exception as e:
                print(f"Error extracting review details on page {page}: {e}")
                # continue # Lanjutkan ke ulasan berikutnya

        # Jeda sejenak antar halaman untuk menghindari pemblokiran
        time.sleep(2) # Sesuaikan durasi jeda jika perlu

    except requests.exceptions.RequestException as e:
        print(f"Error accessing the page {url_saat_ini}: {e}")
        # Jika ada error saat mengakses halaman, mungkin halaman tidak ada atau ada masalah jaringan.
        # Anda bisa memilih untuk melanjutkan ke halaman berikutnya atau berhenti.
        # Saat ini, loop akan berhenti jika ada error RequestException.
        break
    except Exception as e:
        print(f"An unexpected error occurred on page {page}: {e}")
        break # Berhenti jika ada error lain yang tidak terduga

print(f"\nSelesai scraping. Total {len(data_ulasan)} ulasan terkumpul.")

# Buat DataFrame dari data yang terkumpul
df = pd.DataFrame(data_ulasan)

# ... (bagian menampilkan head, info, dan menyimpan ke CSV) ...
```
