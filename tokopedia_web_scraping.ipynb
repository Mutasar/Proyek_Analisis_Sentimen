```python
# 1. Import Library
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
```

```python
# 2. Konfigurasi Chrome Driver
options = Options()
options.add_argument("--headless")  # agar tidak membuka jendela browser
driver = webdriver.Chrome(options=options)
```

```python
# 3. Buka URL
url = "https://www.tokopedia.com/pentolban/review"
driver.get(url)
time.sleep(5)  # Tunggu konten termuat
```

```python
# 4. Scroll (jika perlu muat ulasan lebih banyak)
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(3)
```

```python
# 5. Ambil Data Review
usernames = driver.find_elements(By.CSS_SELECTOR, "span.css-1n5guzk")  # Nama pengguna
ratings = driver.find_elements(By.CSS_SELECTOR, "span.css-1ff3g6v")    # Rating bintang
comments = driver.find_elements(By.CSS_SELECTOR, "p.css-1kdc32b")       # Komentar
```

```python
# 6. Simpan ke List
data = []
for i in range(len(comments)):
    data.append({
        'username': usernames[i].text if i < len(usernames) else "",
        'rating': ratings[i].text if i < len(ratings) else "",
        'comment': comments[i].text
    })
```

```python
# 7. Simpan ke DataFrame
df = pd.DataFrame(data)
print(df.head())
```

```python
# 8. Simpan ke CSV
df.to_csv("tokopedia_reviews.csv", index=False)
```

```python
# 9. Tutup Browser
driver.quit()
```
