```python
import gdown
import pandas as pd
import os

# The Google Drive file ID can be extracted from the URL
# For a URL like: https://drive.google.com/file/d/1GLyhzyN3XNgmC3ftv4Klmvtlt_Dc7XzD/view?usp=sharing
# The file ID is '1GLyhzyN3XNgmC3ftv4Klmvtlt_Dc7XzD'
file_id = '1GLyhzyN3XNgmC3ftv4Klmvtlt_Dc7XzD'
output_filename = 'downloaded_data.xlsx' # Or .csv, .txt, etc., depending on the actual file type

try:
    # Download the file
    gdown.download(id=file_id, output=output_filename, quiet=False)
    print(f"File downloaded to {output_filename}")

    # Now, read the data based on its type
    if output_filename.endswith('.csv'):
        df = pd.read_csv(output_filename)
        print("\nCSV Data:")
        print(df.head())
    elif output_filename.endswith('.xlsx'):
        df = pd.read_excel(output_filename)
        print("\nExcel Data:")
        print(df.head())
    elif output_filename.endswith('.txt'):
        with open(output_filename, 'r', encoding='utf-8') as f:
            content = f.read()
            print("\nText File Content:")
            print(content[:500]) # Print first 500 characters
    else:
        print(f"Unsupported file type for direct reading: {output_filename}. You'll need specific libraries.")

except Exception as e:
    print(f"An error occurred: {e}")
    print("Please ensure the Google Drive file is accessible (e.g., 'Anyone with the link' access) and the file ID is correct.")
finally:
    # Optional: Clean up the downloaded file
    if os.path.exists(output_filename):
        # os.remove(output_filename)
        print(f"\nDownloaded file '{output_filename}' remains for inspection.")
        pass # Keep the file for now for inspection
```
